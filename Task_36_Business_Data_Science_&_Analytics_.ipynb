```python
import pandas as pd
import numpy as np

np.random.seed(42)

# 1. Website Visitors Dataset

n = 1000

website = pd.DataFrame({
    "Visitor_ID": range(1, n + 1),
    "Date": pd.date_range("2026-01-01", periods=n, freq="h"),
    "Location": np.random.choice(
        ["Guntur", "Vijayawada", "Hyderabad", "Chennai", "Bangalore"],
        n
    ),
    "Device": np.random.choice(
        ["Mobile", "Desktop", "Tablet"], n
    ),
    "Marketing_Channel": np.random.choice(
        ["Google", "Instagram", "LinkedIn", "Direct", "Email"], n
    ),
    "Pages_Viewed": np.random.randint(1, 15, n),
    "Session_Minutes": np.random.randint(1, 30, n),
    "Converted": np.random.choice(
        ["Yes", "No"], n, p=[0.3, 0.7]
    )
})

website.to_csv(
    base_path + "/dataset/website_visitors.csv",
    index=False
)


# 2. Internship Applications Dataset

n = 500

internships = pd.DataFrame({
    "Application_ID": range(1, n + 1),
    "Application_Date": pd.date_range("2026-01-01", periods=n, freq="D"),
    "Student_Location": np.random.choice(
        ["Guntur", "Vijayawada", "Hyderabad", "Chennai"], n
    ),
    "Internship_Role": np.random.choice(
        ["Python", "Data Science", "Java", "Web Development", "AI"],
        n
    ),
    "Application_Status": np.random.choice(
        ["Selected", "Rejected", "Pending"], n
    ),
    "Experience_Level": np.random.choice(
        ["Beginner", "Intermediate", "Advanced"], n
    )
})

internships.to_csv(
    base_path + "/dataset/internship_applications.csv",
    index=False
)


# 3. Training Programs Dataset

n = 300

training = pd.DataFrame({
    "Student_ID": range(1, n + 1),
    "Program": np.random.choice(
        ["Python", "Java", "Data Science", "Machine Learning", "SQL"],
        n
    ),
    "Enrollment_Date": pd.date_range("2026-01-01", periods=n, freq="D"),
    "Duration_Weeks": np.random.choice([4, 6, 8, 12], n),
    "Completion_Status": np.random.choice(
        ["Completed", "Ongoing", "Dropped"], n
    ),
    "Rating": np.random.randint(1, 6, n)
})

training.to_csv(
    base_path + "/dataset/training_programs.csv",
    index=False
)


# 4. IT Service Requests Dataset

n = 400

services = pd.DataFrame({
    "Request_ID": range(1, n + 1),
    "Request_Date": pd.date_range("2026-01-01", periods=n, freq="D"),
    "Service_Type": np.random.choice(
        ["Web Development", "Cloud Services", "Software Development",
         "Cybersecurity", "Data Analytics"],
        n
    ),
    "Request_Channel": np.random.choice(
        ["Website", "Email", "Phone", "Social Media"], n
    ),
    "Priority": np.random.choice(
        ["High", "Medium", "Low"], n
    ),
    "Request_Status": np.random.choice(
        ["Completed", "Pending", "In Progress"], n
    ),
    "Satisfaction_Rating": np.random.randint(1, 6, n)
})

services.to_csv(
    base_path + "/dataset/service_requests.csv",
    index=False
)


# 5. Marketing Channels Dataset

n = 200

marketing = pd.DataFrame({
    "Campaign_ID": range(1, n + 1),
    "Month": np.random.choice(
        ["January", "February", "March", "April", "May", "June"], n
    ),
    "Channel": np.random.choice(
        ["Google", "Instagram", "LinkedIn", "Email", "Facebook"], n
    ),
    "Spend": np.random.randint(1000, 10000, n),
    "Impressions": np.random.randint(1000, 20000, n),
    "Clicks": np.random.randint(100, 2000, n),
    "Leads": np.random.randint(10, 500, n),
    "Conversions": np.random.randint(1, 100, n)
})

marketing.to_csv(
    base_path + "/dataset/marketing_channels.csv",
    index=False
)

print("All 5 datasets created successfully!")
```
